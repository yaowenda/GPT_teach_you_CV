## 4 学习Swin时最该抓住的几个点

这一章不再展开细节，只做最后收束：**把 Swin 最值得记住的两个判断压出来。**

### 4.1 它和ViT最本质的区别是什么？

不要把 Swin 理解成“ViT 加窗口”。更准确地说：

- `ViT` 更像在**固定尺度**上做**全局序列建模**
- `Swin` 更像在**局部计算**、**跨窗传播**和**尺度演化**中，逐步构建层次化视觉表示

所以两者最本质的区别有两点：

1. `注意力组织方式不同`
   ViT 偏全局注意力；Swin 偏局部窗口注意力 + shifted window。

2. `特征演化方式不同`
   ViT 更偏固定尺度；Swin 通过 patch merging 形成多 stage、多尺度特征。

一句话记忆：**ViT 更像全局序列 Transformer，Swin 更像层次化视觉 backbone。**


### 4.2 “局部计算 + 跨窗连接 + 层次化表示”该如何整体理解？

把 Swin 整体看成三步就够了：

1. `局部计算`
   先在窗口内做注意力，解决高分辨率下计算太贵的问题。

2. `跨窗连接`
   再用 shifted window 让信息跨窗口传播，避免局部窗口彼此隔离。

3. `层次化表示`
   再用 patch merging 让特征图分辨率下降、语义增强，形成多 stage backbone。

所以 Swin 的整体逻辑不是“直接全局看整张图”，而是：

**先局部看清，再跨窗传播，最后分层汇聚。**

学习时最容易混淆的点只有三个：

- `shifted window` 不是全局注意力，它仍然是局部计算
- `patch merging` 不只是下采样，它还在做语义汇聚和通道重组
- Swin 不是“ViT 加窗口”，而是“层次化视觉 Transformer 主干”

如果整本讲义最后只记一句话，那就记：

**Swin Transformer 的核心，是把“局部高效计算、跨窗口传播、层次化尺度演化”三件事组合起来，做成了一个真正适合视觉任务的 Transformer backbone。**
